```python
import numpy as np
import matplotlib.pyplot as plt
from tabula import read_pdf
import pandas as pd
df = read_pdf(
    'https://www3.akindo-sushiro.co.jp/pdf/menu/allergy.pdf',
    pages='all',
    multiple_tables=True,
    lattice=True,
    force_subprocess=True  # Add this line
)
df
```

```python
df[0].columns
```

```python
# 1. Combine all pages into one large DataFrame
df_combined = pd.concat(df, ignore_index=True)

# 2. Define the exact column names you are looking for
non_halal_cols = ['落\r花\r生', '牛\r肉', '鶏\r肉', '豚\r肉', 'ゼ\rラ\rチ\rン']

# 3. Filter to only include columns that actually exist in the parsed data
# (This prevents KeyErrors if Tabula missed a column name)
existing_cols = [col for col in non_halal_cols if col in df_combined.columns]

# 4. Create a mask: True if ANY of the existing columns have '●' in that row
mask = (df_combined[existing_cols] == '●').any(axis=1)

# 5. Extract the menu names, drop empty values, and get unique items
takhalal = df_combined.loc[mask, 'メニュー名称'].dropna().unique().tolist()

print(takhalal)
```

```python
for i in range(np.size(takhalal)):
    print(takhalal[i])
```

```python
# Open a text file in write mode ('w') with UTF-8 encoding
with open('sushiro_non_halal_menu.txt', 'w', encoding='utf-8') as file:
    for item in takhalal:
        # Write each item followed by a newline character (\n)
        file.write(f"{item}\n")

print("File saved as sushiro_non_halal_menu.txt")
```
